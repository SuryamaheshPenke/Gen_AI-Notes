## **PART 1 - The LLM Landscape: ChatGPT, Claude, Gemini**

In late 2022, OpenAI released ChatGPT and made large language models accessible to the public for the first time. Within months, Google and Anthropic shipped their own general-purpose models. By 2024, all three had become serious production tools used by engineers, analysts, writers, and product teams.

The problem: they look identical from the outside. All three take text in and return text out. All three can answer questions, write code, summarize documents. So why does it matter which one you use?

It matters because the differences are real and consequential once you move past toy prompts. Context window sizes differ by 10x or more. Each model was trained with different priorities - one is tuned for instruction-following, another for reasoning, another for integrating into Google's product ecosystem. Pricing varies widely across input/output tokens and model tiers. Choosing the right model for a task is an engineering decision, not a preference.

This part builds the mental model for that decision.

---

### ChatGPT (OpenAI)

**Simple**
ChatGPT is the interface; GPT-4o is the model underneath it. OpenAI is the most widely deployed LLM provider. Its models are strong generalists - they handle code, reasoning, and creative tasks well. The API is the industry default, meaning most tutorials and tooling assume OpenAI.

**Technical**
OpenAI's GPT-4o is a multimodal, decoder-only transformer that accepts text, images, and audio. It supports function calling, JSON mode, and structured outputs via `response_format`. The `gpt-4o` context window is 128K tokens. The `o1` and `o3` model families use test-time compute scaling (extended chain-of-thought reasoning) for higher accuracy on math and logic tasks at higher cost. Pricing is per input/output token with a separate tier for cached input tokens.

---

### Claude (Anthropic)

**Simple**
Claude is Anthropic's model family. It is known for following long, complex instructions accurately, handling large documents, and producing clean, readable prose. Anthropic was founded specifically to work on AI safety, so Claude's behavior is shaped by that - it is less likely to drift from instructions mid-response and more conservative about harmful outputs.

**Technical**
Claude 3.5 Sonnet and Claude 3.7 Sonnet are the current production models. Claude's context window is 200K tokens, the largest among the three providers. The API uses a `messages` format with a separate top-level `system` parameter for system prompts. Claude supports tool use (function calling), streaming, vision, and extended thinking mode (test-time compute scaling analogous to OpenAI's o1). It does not support JSON mode by default but can be reliably prompted to return structured output or constrained via tool schemas.

---

### Gemini (Google DeepMind)

**Simple**
Gemini is Google's model family. Its standout trait is context window size - Gemini 1.5 Pro supports up to 1 million tokens, which means you can feed it entire codebases, long books, or hours of video. It is natively multimodal from the ground up, not retrofitted.

**Technical**
Gemini 1.5 Pro has a 1M token context window; Gemini 1.5 Flash targets lower latency and cost. The models are built on a Mixture-of-Experts architecture and trained natively on text, images, audio, and video. Access is via Google AI Studio (prototyping) or Vertex AI (production). Function calling and structured output are supported. The `google-generativeai` Python SDK is the primary client library.

### When to Use Which

**Simple**
Pick GPT-4o when you want the most ecosystem support - tutorials, libraries, and examples will almost always target OpenAI first. Pick Claude when your task involves long documents or when precise instruction-following matters more than raw speed. Pick Gemini when you need to process huge inputs (a full codebase, a long video) or when you are already inside Google infrastructure.

**Technical**
Model selection is a function of: (1) context window requirement, (2) output type (structured JSON, code, prose), (3) cost per token at your expected volume, (4) latency requirements, (5) compliance and data residency. For production systems, run the same prompt across models and evaluate outputs against your metric before locking in a provider. Most serious systems use multiple models - a fast cheap model for classification, a larger model for generation, a reasoning model for evaluation.

## **PART 2 - Coding Assistants: GitHub Copilot vs Cursor**

LLMs became useful for code before most people used them for anything else. GitHub Copilot launched in 2021 as an IDE plugin trained on public code. It autocompleted lines and functions inside the editor. That was useful, but limited - you could not explain a task in English and get working code back.

Cursor arrived later as a fork of VS Code built around chat-first, context-aware code generation. It represents a different model of how developers interact with AI: not just autocompletion, but a full agent that can read your entire codebase, make multi-file edits, and run commands.

The difference matters because one augments how you already work; the other asks you to change how you work.

---

### GitHub Copilot

**Simple**
Copilot lives inside your existing editor (VS Code, JetBrains, Neovim) as a plugin. It watches what you type and suggests completions - single lines, full functions, entire test files. You stay in your normal workflow; Copilot adds suggestions underneath your cursor.

**Technical**
Copilot uses a fine-tuned version of OpenAI's Codex/GPT-4 family. It sends a context window of surrounding code (file content, open tabs, recent edits) to the model and returns completions inline. Copilot Chat adds a conversational interface within the editor. As of 2024, Copilot supports multi-file context awareness via Copilot Workspace, which can plan and execute changes across a repo given a plain-English issue description. Access requires a GitHub subscription (Individual or Business tier).

---

### Cursor

**Simple**
Cursor is a full editor, not a plugin. You open Cursor instead of VS Code. The editor is built around a chat panel that knows your codebase - you can ask "why does this function fail on empty input" and it reads the actual code to answer. You can tell it "add pagination to this API endpoint" and it edits multiple files at once.

**Technical**
Cursor is built on VS Code (open-source core) with an integrated LLM layer. It indexes your codebase and embeds it for retrieval, so the model's context window always contains the relevant code for your query. The `Composer` feature (now `Agent` mode) supports multi-file, multi-step edits with an approval flow. Cursor routes queries to multiple backends (GPT-4o, Claude, its own fine-tuned models) depending on the task. It supports `.cursorrules` files for project-level instructions that persist across sessions - equivalent to a persistent system prompt.

---

### Side-by-Side Comparison

| Dimension | GitHub Copilot | Cursor |
|---|---|---|
| What it is | IDE plugin | Standalone editor (VS Code fork) |
| Primary interaction | Inline autocompletion + chat | Chat-first, multi-file agent |
| Codebase awareness | Open tabs + recent edits | Full repo indexing + semantic search |
| Multi-file edits | Copilot Workspace (limited) | Native (Composer/Agent mode) |
| Persistent instructions | Not natively | `.cursorrules` file |
| Setup | Install plugin in existing editor | Download and use instead of VS Code |
| Cost | GitHub subscription | Cursor subscription (free tier available) |
| Best fit | Staying in existing workflow, autocompletion speed | New projects, complex refactors, context-heavy tasks |

**Simple**
If you already use VS Code and want AI help without changing anything, use Copilot. If you are starting a new project and want an AI that understands your whole codebase from day one, use Cursor.

**Technical**
The architectural difference is retrieval. Copilot's context is bounded by what fits in the editor's open windows. Cursor maintains a vector index of the repository, retrieving relevant code snippets into the model's context on each query. This makes Cursor significantly more effective on tasks that require understanding code spread across many files.

## **PART 3 - Real-World Applications Across Roles**

LLMs are general tools, but the use cases that actually save time are specific. Here is a concrete map of where AI tools produce measurable output in a typical software and product organization.

---

### Across Software Engineering

**Simple**
Engineers use LLMs to write boilerplate faster, debug code by explaining it to the model, generate tests for existing functions, and understand unfamiliar codebases. The model is not writing production code autonomously - it is removing the tedious parts so the engineer can focus on decisions.

**Technical**
High-value engineering use cases: (1) Code generation - write a function signature + docstring, get a working implementation to edit. (2) Test generation - paste a function, ask for pytest unit tests with edge cases. (3) Code review assistance - explain what this PR does, what could go wrong. (4) Documentation - generate inline docstrings or a README from code. (5) Debugging - paste the stack trace and the relevant code, ask what is wrong. (6) Migration - convert Python 2 to Python 3, or Express to FastAPI. Each use case benefits from a different prompting approach.

---

### Across Product and Analysis

**Simple**
Product teams use LLMs to synthesize user research faster, draft PRDs from bullet points, generate user story variations, and analyze survey data. Analysts use them to write SQL from plain English, summarize dashboards, and explain statistical outputs to non-technical stakeholders.

**Technical**
High-value product/analytics use cases: (1) PRD drafting - give the problem statement and constraints, get a structured document to edit. (2) User research synthesis - paste interview transcripts, ask for themes and quotes. (3) SQL generation - describe the metric and schema, get a query. (4) Competitive analysis - structured prompts against public documents. (5) A/B test interpretation - paste results, ask for a plain-language summary with caveats. The quality of output scales directly with the specificity of the prompt and the quality of the context provided.

---

### Across Content and Communication

**Simple**
Writing teams use LLMs to go from rough notes to polished drafts faster, adapt content for different audiences, and generate variations for testing. The model handles structure and phrasing; the human handles accuracy, tone, and judgment.

**Technical**
High-value content use cases: (1) Long-form drafting from outlines. (2) Tone adaptation - rewrite this for a technical vs non-technical audience. (3) Email and Slack message drafting from bullet points. (4) Summarization - paste a 20-page document, get a 5-point summary. (5) Localization prep - flag idioms and culture-specific references before human translation. LLMs are not reliable fact sources; they are reliable structure and language tools.

## **PART 4 - No-Code AI Tools: Bolt, Lovable, V0, n8n**

LLMs did not just improve what engineers build - they enabled people who do not code to build software directly. Four tools are worth knowing.

**Bolt** (bolt.new) generates a running full-stack web application from a plain-English description and executes it in-browser using a sandboxed Node.js environment (WebContainers). Use it when you need a working prototype in under 10 minutes without any local setup.

**Lovable** (lovable.dev) builds complete web applications - front end, back end, database, and authentication - from a conversation. Output is a real GitHub repository with a React + TypeScript frontend and Supabase backend. Use it when you want an MVP you can actually deploy and hand off to an engineering team.

**V0** (v0.dev, by Vercel) generates production-quality React + Tailwind CSS UI components from a description. It does not build full apps - it builds the individual screens and components you drop into an existing project. Use it when you know what a page should look like but do not want to write the Tailwind yourself.

**n8n** is an open-source workflow automation tool (like Zapier, but self-hostable with an LLM layer). You connect nodes on a canvas - triggers, data transformations, API calls, and AI Agent nodes that can call tools and loop. Use it when you want to automate a multi-step business process that involves both external APIs and LLM reasoning, without writing infrastructure code.

### **Demo 2 - Call OpenAI's GPT-4o**

This sends one prompt to GPT-4o and prints the response. One concept: making a basic chat completion call.

```python
from openai import OpenAI

client = OpenAI()  # reads OPENAI_API_KEY from environment

PROMPT = "Explain what a transformer model is in exactly three sentences. Be precise."

response = client.chat.completions.create(
    model="gpt-4o",
    messages=[
        {"role": "user", "content": PROMPT}
    ],
    max_tokens=200
)

gpt_output = response.choices[0].message.content

print("=== GPT-4o ===")
print(gpt_output)
print()
print(f"Tokens used - input: {response.usage.prompt_tokens}, output: {response.usage.completion_tokens}")
```

What to look at: the three sentences GPT-4o chose, what it emphasized, and the token counts. `prompt_tokens` is what you sent; `completion_tokens` is what the model generated. Token counts drive cost.

### **Demo 3 - Call Anthropic's Claude with the same prompt**

Same prompt, different provider. The API structure differs: Anthropic separates the system prompt from the user message at the top level.

```python
import anthropic

client_a = anthropic.Anthropic()  # reads ANTHROPIC_API_KEY from environment

# SAME prompt as Demo 2
PROMPT = "Explain what a transformer model is in exactly three sentences. Be precise."

response_a = client_a.messages.create(
    model="claude-3-5-sonnet-20241022",
    max_tokens=200,
    messages=[
        {"role": "user", "content": PROMPT}
    ]
)

claude_output = response_a.content[0].text

print("=== Claude 3.5 Sonnet ===")
print(claude_output)
print()
print(f"Tokens used - input: {response_a.usage.input_tokens}, output: {response_a.usage.output_tokens}")
```

What to look at: compare the sentences to GPT-4o's output. Same instruction, same three-sentence constraint - does Claude interpret the task the same way? Does it use different terminology or emphasis? This is the calibration exercise: same prompt, observe the difference.

### **Demo 4 - Side-by-Side Comparison**

Print both outputs together for direct comparison. Also shows the structural difference between the two API response objects.

```python
# Side-by-side output
# Requires Demo 2 and Demo 3 to have run first.

print("PROMPT")
print("-" * 60)
print(PROMPT)
print()

print("GPT-4o")
print("-" * 60)
print(gpt_output)
print()

print("Claude 3.5 Sonnet")
print("-" * 60)
print(claude_output)
print()

# API structure difference
print("API STRUCTURE NOTE")
print("-" * 60)
print("OpenAI:    response.choices[0].message.content")
print("Anthropic: response.content[0].text")
print()
print("OpenAI token field:    response.usage.prompt_tokens / completion_tokens")
print("Anthropic token field: response.usage.input_tokens / output_tokens")
```

What to look at: the output differences, and the field names at the bottom. When you build a system that supports multiple providers, you need to normalize these response structures into a common format. The difference in field names is small but breaks your code if you forget it.

# **Key Takeaways**

1. GPT-4o, Claude, and Gemini are all strong general-purpose models. The choice depends on context window size needed, instruction fidelity requirements, and cost at your expected volume - not on brand preference.

2. Model specs change fast. The comparison table in this notebook is a snapshot. Always verify context window sizes, pricing, and model availability on the provider's documentation page before building a production system.

3. Copilot augments your existing editor with inline suggestions. Cursor replaces your editor with an agent that indexes your codebase. The architectural difference is retrieval depth, not just interface.

4. No-code tools like Bolt, Lovable, V0, and n8n lower the cost of prototyping to near zero. They do not eliminate the need for engineering judgment - they shift where that judgment is applied.

5. The OpenAI and Anthropic APIs have different response structures. `response.choices[0].message.content` vs `response.content[0].text`. Normalizing these is the first thing you do when building a multi-provider system.

6. Same prompt, different models - the outputs differ in ways that matter for specific tasks. Running both is fast and cheap. Evaluate before you commit to a provider.

7. LLMs are most valuable when paired with specific, well-scoped tasks. The broader and vaguer the prompt, the less differentiated the output across models.

**1.** Gemini 1.5 Pro is the only viable option. A 500-page legal document at roughly 250 words per page is approximately 125,000-150,000 tokens - this exceeds GPT-4o's 128K limit but fits within Gemini's 1M window. Architecturally, the alternative is chunking across multiple calls, which loses cross-document context and adds complexity.

**2.** OpenAI wraps the response text inside `choices[0].message.content`. Anthropic wraps it inside `content[0].text`. The field names and nesting structures are different. This matters because if you write code that works for one provider and then add the other, the extraction line will break - you need provider-specific logic or a normalization layer to handle both. We will write that normalization in Class 18.

**3.** GitHub Copilot sees only the open file and a few adjacent files. Cursor indexes the entire codebase into a vector store and retrieves relevant context from across all files before generating. This solves multi-file refactoring - Copilot loses context when changes span multiple files; Cursor can reason about the full project structure.

Use V0 when you need a component for an existing project. Use Bolt when you want a fully working prototype without setting up a local environment.

**6.** Different models have different training distributions and default vocabularies, so the same concept will be phrased with different terminology by default. For consistent output regardless of model, define the exact terminology and phrasing in the prompt itself - do not rely on the model's defaults.

**7.** A standard n8n node calls an LLM once and passes the result forward - a fixed, deterministic step.
